````python
import pymupdf

document = pymupdf.open(r"C:\Users\rolex\OneDrive\Desktop\RAG\data\table.pdf")


text = document[0].get_text()

text
````

````python
table = document[0].find_tables()

print(table)

table_data = table[0].extract()

headers = table_data[0]
rows = table_data[1:]

markdown_table = "| " + " | ".join(headers) + " |\n"

markdown_table+= "| " + " | ".join(["---"]*len(headers)) + " |\n"

for row in rows:
    markdown_table += "| " + " | ".join(row) + " |\n"

print("\nMarkdown table:")

print(markdown_table)
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=50)

chunks = splitter.split_text(markdown_table)

chunks
````

````python
from langchain_core.documents import Document

documents = []

for chunk in chunks:
    documents.append(
        Document(
            page_content=chunk,
            metadata={
                "source":"table.pdf",
                "type":"table",
                "page":1
            }
        )
    )    
documents
````
