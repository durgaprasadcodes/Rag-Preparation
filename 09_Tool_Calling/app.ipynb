# Tool Calling

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
@tool(description="Add two integers and return their sum.")
def add(a: int, b: int) -> int:
    print("add tool called")
    return a + b


@tool(description="Multiply two integers and return their product.")
def multiply(a: int, b: int) -> int:
    print("multiply tool called")
    return a * b


@tool(description="Divide the first number by the second and return the result.")
def divide(a: float, b: float) -> float:
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    print("divide tool called")
    return a / b
```

```python
load_dotenv()
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
llm = ChatGroq(model="qwen/qwen3.8-27b", temperature=0, api_key=GROQ_API_KEY)
```

```python
tools = [add, multiply, divide]
```

```python
agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt="""
    You are a helpful mathematics tutor. Use the available tools for every arithmetic operation they support; do not calculate those operations mentally.
    For multi-step calculations, call the tools in sequence and use each result as the next input.
    If a requested operation has no available tool, say so clearly instead of guessing.
    """,
)
```

```python
response = agent.invoke(
    {
        "messages": [
            {"role": "user", "content": "show how to add 100 and 20 then multiply the result with 40"}
        ]
    }
)

print(response["messages"][-1].content)
```
