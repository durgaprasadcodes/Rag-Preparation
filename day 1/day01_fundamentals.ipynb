# 1. What is RAG?

- RAG (Retrieval-Augmented Generation) is a technique where an LLM retrieves relevant information from an external knowledge source and uses that information to generate an answer.

**The Flow is :**

```code

User Question
      ↓
Retrieve relevant information
      ↓
Give information + question to LLM
      ↓
Generate Answer

```

#### Example

**Suppose you have a 500-page company documentation PDF.**

##### User asks:

**"What is the company's leave policy?"**

**RAG will:**

- Search the PDF for relevant information.
- Retrieve the chunks containing the leave policy.
- Send those chunks + the question to the LLM.
- LLM generates the answer based on the retrieved content.
- Simple definition

**RAG = Retrieval + Generation**

> Retrieval → Find relevant information.
> Augmented → Add that information to the prompt.
> Generation → LLM generates the final answer.

### 2. Why RAG instead of Fine-Tuning?

- RAG and fine-tuning solve different problems.

**Fine-tuning**

- Fine-tuning modifies the model's parameters by training it on additional examples.

```code
Base Model
    ↓
Training Data
    ↓
Fine-Tuning
    ↓
Modified Model
```
**RAG**

- RAG keeps the model unchanged and provides external information at runtime.

```code
Base Model
    +
External Knowledge
    ↓
RAG
    ↓
Answer
```

#### Why use RAG?
| **RAG**| **Fine-tuning**|
|:--|--:|
| Adds external knowledge   |  	Changes model behavior|
| Knowledge can be updated easily |	Updating knowledge requires retraining |
| Good for private documents   |	Good for style/behavior/task adaptation |
| Can provide source context |	Knowledge becomes part of model weights|
| Usually easier to maintain |	Training can be expensive|
| Good for Q&A over documents |	Good for specialized behavior|

### 3. What are Embeddings?

- An embedding is a numerical representation of data that captures its semantic meaning.

**For example:**

```code
"I love Python"
        ↓
Embedding Model
        ↓
[0.21, -0.54, 0.73, 0.11, ...]
```

The numbers are called a vector.

Why do we need embeddings?

**Computers don't directly understand that:**

```code

"I love Python"

and

"Python is my favorite programming language"

```

have similar meanings.

An embedding model converts both into vectors.

**Conceptually**:

```code
"I love Python"
        ↓
[0.21, 0.43, -0.12, ...]

"Python is my favorite programming language"
        ↓
[0.20, 0.45, -0.10, ...]
```

- These vectors should be close to each other in vector space.

**Embeddings can represent**:
- Text
- Images
- Audio
- Documents
- Code
- Other data

**In RAG**:

Embeddings are mainly used to convert:

```code
Documents → Vectors
User Query → Vector
```

Then we compare the vectors to find relevant documents.

### 4. What is Semantic Search?

- Semantic search searches based on the meaning of the query rather than only matching exact words.

**Traditional keyword search**

**Query**:

`"How can I reset my password?"`

- It primarily looks for matching words such as:

1.reset<br>
2.password<br>
3.Semantic search<br>

- It can understand that even though many words are different

```code
"I forgot my login credentials. How do I access my account?"

is related to:

"How can I reset my password?"
```


**How does it work?**:
```code
User Query
    ↓
Embedding Model
    ↓
Query Vector
    ↓
Compare with document vectors
    ↓
Find semantically similar documents
Simple definition
```

> Semantic search finds information based on meaning rather than exact keyword matching.

### 6. What is a Context Window?

- A context window is the maximum amount of information an LLM can process in a single request.

- It is generally measured in tokens.

##### For example, conceptually:

`Context Window = 128K tokens`

> means the model can process up to approximately 128,000 tokens in its context, subject to the specific model/API limits.

**In RAG**

##### Suppose your PDF contains:

`500,000 tokens`

- You cannot necessarily send the entire PDF to the model.

**Instead**:

```code
500,000-token document
        ↓
Chunking
        ↓
Many smaller chunks
        ↓
Retrieve relevant chunks
        ↓
Send only relevant chunks to LLM
```

>Context window = the amount of information an LLM can consider in one request.

### 7. What is Hallucination?

- An LLM hallucination occurs when the model generates information that is incorrect, unsupported, or fabricated, while presenting it as if it were factual.

*Example*

**User**:

`"What is the leave policy in our company?"`

- The model doesn't have the company's handbook.

*It might respond*:

`"Employees receive 25 days of annual leave."`

- But the actual policy might say 20 days.

That's a **hallucination.**

#### Why can hallucinations happen?

- LLMs generate responses based on learned patterns and the context provided to them. They don't inherently guarantee that every generated statement is factually correct.

#### How can RAG help?

- RAG provides relevant external context:

```code
Question
   ↓
Retrieve company policy
   ↓
Relevant document:
"Employees receive 20 days..."
   ↓
LLM
   ↓
Answer based on retrieved context
```

- This can reduce hallucinations, but it does not eliminate them completely.

- Poor retrieval can also lead to poor answers.

### 8. What is Similarity Search?

- Similarity search finds vectors that are mathematically closest to a given query vector.

*Suppose we have*:

**Query**:
`"How do I reset my password?"`

**Its embedding might be**:

`Q = [0.2, 0.8, 0.1]`

**Our database contains**:

```code
Document A → [0.21, 0.79, 0.11]
Document B → [0.90, 0.10, 0.40]
Document C → [0.19, 0.81, 0.12]
```

*Documents A and C are closer to the query.*

**So**:

```code
Query
  ↓
[0.2, 0.8, 0.1]
  ↓
Similarity Search
  ↓
Document A
Document C

```
#### Common similarity/distance measures
- Cosine similarity
- Euclidean distance
- Dot product

>For many text-embedding systems, cosine similarity is commonly used.

#### Simple definition

**Similarity search** = `finding the vectors that are most similar to the query vector.`

### 9. Semantic Search vs Similarity Search

- They are closely related, but they describe different levels of the process.

`The easiest way to remember`:

**Semantic Search** =   `What you are trying to achieve` <br>
**Similarity Search** = `How you find the relevant vetors`

**1. Semantic Search**

- Semantic search means searching for information based on the meaning/intention of the query rather than just exact keyword matches.

`Example`

*User searches*:

`"How can I recover my account?"`

*A semantic search system can find*:

`"Steps to reset your password"`

even though the words recover and reset are different.

```code
User Query
    ↓
Understand meaning
    ↓
Find semantically relevant information
    ↓
Relevant documents
```
> So semantic search is a search approach.


**2. Similarity Search**

- Similarity search means finding data points whose representations are mathematically similar to a query representation.

*In an embedding-based system*:

```code
Query
  ↓
Embedding
  ↓
[0.21, 0.73, 0.14, ...]
  ↓
Compare with stored vectors
  ↓
Find closest vectors
```
*For example*:

```code 
Query Vector      → [0.21, 0.73, 0.14]

Document A        → [0.20, 0.72, 0.15]  ← Very similar
Document B        → [0.81, 0.12, 0.63]
Document C        → [0.19, 0.70, 0.16]  ← Very similar
```
*This comparison can use*:

- Cosine similarity
- Dot product
- Euclidean distance

>So similarity search is a retrieval mechanism.

### 10. How They Work Together in RAG


```code
User:

"How can I recover my account?"
          ↓
   Embedding Model
          ↓
    Query Vector
          ↓
    Similarity Search
          ↓
Find vectors closest to query
          ↓
Relevant document chunks
          ↓
      LLM
          ↓
      Answer
```

All These Process Is Called **Semantic Search**

> Similarity search can be the mechanism/process used to implement semantic search.


# 11. Understand the Complete RAG Pipeline

- This is the most important part of Day 1. 🔥

*A typical RAG system has two major phases* :

```code
        RAG SYSTEM
            │
     ┌──────┴──────┐
     ↓             ↓
INDEXING        RETRIEVAL
 PHASE            PHASE
```


## Phase 1 — Indexing
- Query Extraction and chucking ,Embedding Transformation
- This happens before the user asks a question.

### Step 1 — **Load Documents**

*Suppose we have*:

`company_handbook.pdf`

We load the document.

```code
PDF
 ↓
Document Loader
 ↓
Document Text
```

### Step 2 — **Split into Chunks**

- Large documents are divided into smaller pieces.

```code
Large Document
      ↓
┌──────────────┐
│ Chunk 1      │
├──────────────┤
│ Chunk 2      │
├──────────────┤
│ Chunk 3      │
├──────────────┤
│ Chunk 4      │
└──────────────┘
```
**Why** ?

- Because sending an entire large document to the LLM is inefficient and may exceed the context window.

### Step 3 — **Create Embeddings**

- Each chunk is converted into a vector.

```code
Chunk 1
   ↓
Embedding Model
   ↓
Vector 1

Chunk 2
   ↓
Embedding Model
   ↓
Vector 2
```

### Step 4 — **Store in Vector Database**

- The chunks and their embeddings are stored.

```code
Chunk 1 → Vector 1
Chunk 2 → Vector 2
Chunk 3 → Vector 3
Chunk 4 → Vector 4
```

Inside a vector database.

## Phase 2 — Retrieval + Generation

- Now the user asks a question.

### Step 5 — **User Query**

`"How many annual leave days do employees receive?"`

### Step 6 — **Convert Query into Embedding**

- The query is converted into a vector.

```code
User Query
    ↓
Embedding Model
    ↓
Query Vector
```
### Step 7 — **Similarity Search**

- The query vector is compared against document vectors.

```code
Query Vector
      ↓
Vector Database
      ↓
Similarity Search
      ↓
Top-K Relevant Chunks
```

*For example*:

```code
Chunk 24 → 0.92 similarity
Chunk 18 → 0.87 similarity
Chunk 31 → 0.81 similarity
```

> The system retrieves the most relevant chunks.

### Step 8 — **Build the Prompt**

*Now we combine*:

```code
User Question
       +
Retrieved Context
       ↓
     Prompt
```

**Conceptually**:

*Context*:
-Employees receive 20 days of annual leave per year.

*Question*:
- How many annual leave days do employees receive?

### Step 9 — **Send to LLM**

```code
Prompt
  ↓
LLM
  ↓
Generated Answer
```

>The LLM uses the retrieved context to formulate the response.

### Step 10 — **Final Answer**

*User*:
- How many annual leave days do employees receive?

*RAG*:
- Employees receive 20 days of annual leave per year.
