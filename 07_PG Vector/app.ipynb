# PG Vector Tutorial

```python
import os
from dotenv import load_dotenv

from langchain_postgres import PGVector
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain.chat_models import init_chat_model

load_dotenv()

# =========================
# Environment Variables
# =========================

GROQ_API_KEY = os.getenv("GROQ_API_KEY")
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
DATABASE_URL = os.getenv("DATABASE_URL")


# =========================
# Groq Chat Model
# =========================

model = init_chat_model(
    model="qwen/qwen3.8-27b", model_provider="groq", api_key=GROQ_API_KEY
)


# =========================
# Gemini Embedding Model
# =========================

embedding_model = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001", google_api_key=GEMINI_API_KEY
)


# =========================
# PGVector
# =========================

vector_store = PGVector(
    embeddings=embedding_model,
    collection_name="learnx_documents",
    connection=DATABASE_URL,
    use_jsonb=True,
)


print("PGVector connected successfully!")
```

```python
from langchain_core.documents import Document

documents = [
    Document(
        page_content="Python is a high-level programming language.",
        metadata={"topic": "python"},
    ),
    Document(
        page_content="FastAPI is a Python framework for building APIs.",
        metadata={"topic": "fastapi"},
    ),
    Document(
        page_content="PostgreSQL is a powerful relational database.",
        metadata={"topic": "database"},
    ),
]
```

```python
vector_store.add_documents(documents)

print("Documents added successfully!")
```

```python
query = "What is FastAPI?"

results = vector_store.similarity_search(query, k=3)

context = "\n\n".join(document.page_content for document in results)

prompt = f"""
Answer the question using the context below.

Context:
{context}

Question:
{query}
"""

response = model.invoke(prompt)

print(response.content)
```
