# 1. PDF Parsing
*What is it?*

**PDF** parsing means:

- Reading a PDF and extracting its text/content so that our RAG system can process it.

- For LangChain, PyMuPDFLoader is a very useful choice.

**Install**
```py
pip install langchain langchain-community pymupdf docx2txt

```

# PDF Loading

````python
from langchain_community.document_loaders import PyMuPDFLoader

loader = PyMuPDFLoader(r"C:\Users\rolex\OneDrive\Desktop\RAG\data\sample10.pdf")

documents = loader.load()

print(len(documents))
print(documents[0].page_content[:100])
print(documents[0].metadata)
````

# DOCX Loading

````python
from langchain_community.document_loaders import Docx2txtLoader

loader = Docx2txtLoader(r"C:\Users\rolex\OneDrive\Desktop\RAG\data\docx_file.docx")

document = loader.load()

print(len(document))
print(document[0].page_content[:100])
print(document[0].metadata)
````

# TXT Loading

````python
from langchain_community.document_loaders import TextLoader

loader = TextLoader(r"C:\Users\rolex\OneDrive\Desktop\RAG\data\txt_file.txt")

document = loader.load()

print(len(document))
print(document[0].page_content[:72])
print(document[0].metadata)
````

# CSV Loading

````python
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(r"C:\Users\rolex\OneDrive\Desktop\RAG\data\csv_file.csv")

document = loader.load()

print(len(document))

for doc in document:
    print(doc.page_content)
````
