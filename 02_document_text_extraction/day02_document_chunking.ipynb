````python
pdf_path = r"C:\Users\rolex\OneDrive\Desktop\RAG\data\sample10.pdf"

doc = fitz.open(pdf_path)

print("Number of pages:", len(doc))
````

### Step 3 — Extract the Entire PDF


````python
all_text = ""

for page in doc:
    all_text += page.get_text()

all_text[:3000]
````

````python
from langchain_core.documents import Document

documents  = []

for page_num,page in enumerate(doc):
    text = page.get_text()
    
    document = Document(
        page_content=text,
        metadata={
            "source":pdf_path,
            "page":page_num+1
        }
    )
    documents.append(document)
    
documents[0]
````

### Step 5 — Create LangChain Documents

You now have to understand this structure.

```text
Document
├── page_content
│      └── actual text
│
└── metadata
       ├── source
       └── page

```

- This is extremely important for production RAG.

### Step 6 — Inspect Your Documents

````python
len(documents)
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)

chunks
````
