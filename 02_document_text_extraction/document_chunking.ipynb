# RecursiveCharacterTextSplitter
- Split text using a hierarchy of separators while trying to keep meaningful pieces together.

*For example*:

```code
Paragraph
   ↓
Sentence
   ↓
Word
   ↓
Character

```
Conceptually, it tries separators such as:

```code
"\n\n"   → paragraph
"\n"     → line
" "      → word
""       → character
```

````python
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = PyMuPDFLoader(r"C:\Users\rolex\OneDrive\Desktop\RAG\data\sample.pdf")

document = loader.load()

print(len(document),document[0].metadata)

splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)

chunks = splitter.split_documents(document)

chunks
````

# Parent-Child Chunking 

- This is one of the most useful concepts for advanced RAG.

*The basic idea:*

- Store/search using small child chunks, but return the larger parent chunk for context.

> This solves a major RAG problem.
