# 1. What exactly is an embedding?

- An embedding is a numerical representation of data—usually text—where the numbers encode useful semantic relationships.

```code
"Python is a programming language"
              ↓
        Embedding Model
              ↓
     [0.21, -0.73, 0.42, ...]
              ↓
       Vector representation
              ↓
Compare with another vector
              ↓
   Semantic similarity

```

````python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

embedding = embedding_model.encode("I love learning Python")

print(type(embedding))
print(embedding.shape)
print(embedding[:10])
````
